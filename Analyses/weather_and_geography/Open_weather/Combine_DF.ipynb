```python
import pandas as pd

# Load datasets
weather_df = pd.read_csv("aggregated_30d_with_zscores.csv")
pca_df = pd.read_csv("PCA_Spore_Scores_with_Ecology_Hand_Labeled.csv")

```

```python
# STEP 1: Create species_clean column in weather_df
# Convert 'A_abrupta' -> 'amanita_abrupta'
weather_df["species_clean"] = weather_df["taxa_name"].str.replace("A_", "amanita_", regex=False).str.lower()

# STEP 2: Merge weather data with PCA data
# Left join: keep all weather records, enrich with PCA/ecology info
merged_df = pd.merge(weather_df, pca_df, on="species_clean", how="left")

# STEP 3: Save to new CSV
merged_df.to_csv("merged_weather_pca.csv", index=False)

print("✅ Merge complete. Output saved to 'merged_weather_pca.csv'.")
```
